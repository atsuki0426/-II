```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

url = "https://www.nikkansports.com/baseball/professional/score/2026/pl2026051004.html"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept-Language": "ja-JP,ja;q=0.9",
}

res = requests.get(url, headers=headers, timeout=10)
res.raise_for_status()

soup = BeautifulSoup(res.text, "html.parser")
text = soup.get_text("\n", strip=True)

lines = text.splitlines()

# 打席結果として出やすい表記
result_words = [
    "遊ゴ", "二ゴ", "一ゴ", "三ゴ", "投ゴ", "捕ゴ",
    "遊飛", "二飛", "一飛", "三飛", "投飛", "捕飛",
    "左飛", "中飛", "右飛",
    "遊直", "二直", "一直", "三直", "投直",
    "三振", "空三振", "見三振",
    "四球", "死球",
    "左安", "中安", "右安",
    "左２", "中２", "右２",
    "左本", "中本", "右本",
    "中犠", "右犠", "左犠",
    "遊併", "二併", "三併", "一併", "投併"
]

rows = []

# まずページ内で、打者名と打席結果が同じ行に出ている行を確認する
for line in lines:
    if any(w in line for w in result_words):
        rows.append({"raw_line": line})

df = pd.DataFrame(rows)
df.to_csv("nikkansports_raw_batting_lines.csv", index=False, encoding="utf-8-sig")

print(df.head(30))
print("保存しました: nikkansports_raw_batting_lines.csv")
```
