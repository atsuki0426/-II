```python
import pandas as pd
from pathlib import Path

# CSV読み込み
df = pd.read_csv("中間発表.csv")

# 数値列を整える
numeric_cols = ["RE24", "WPA", "打点"]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)

# 本塁打列を作る
if "打席結果" in df.columns:
    df["本塁打"] = df["打席結果"].astype(str).str.contains("本塁打|ホームラン|HR").astype(int)
else:
    df["本塁打"] = 0

# summaryを作成
summary = (
    df.groupby(["球団", "打者名"], as_index=False)
    .agg(
        打席数=("打者名", "count"),
        RE24合計=("RE24", "sum"),
        WPA合計=("WPA", "sum"),
        打点合計=("打点", "sum"),
        本塁打合計=("本塁打", "sum"),
    )
)

summary["1打席あたりRE24"] = summary["RE24合計"] / summary["打席数"]
summary["1打席あたりWPA"] = summary["WPA合計"] / summary["打席数"]

summary = summary.sort_values("RE24合計", ascending=False)

display(summary.head(10))
```

```python
re24_top10 = summary.sort_values("RE24合計", ascending=False).head(10)

display(
    re24_top10[
        ["球団", "打者名", "打席数", "RE24合計", "WPA合計", "打点合計", "本塁打合計", "1打席あたりRE24"]
    ]
)
```

```python
display(
    summary.sort_values("RE24合計", ascending=False)
    .head(10)[
        ["球団", "打者名", "打席数", "RE24合計", "WPA合計", "打点合計", "本塁打合計", "1打席あたりRE24"]
    ]
)
```

```python
wpa_top10 = summary.sort_values("WPA合計", ascending=False).head(10)

display(
    wpa_top10[
        ["球団", "打者名", "打席数", "RE24合計", "WPA合計", "打点合計", "本塁打合計", "1打席あたりWPA"]
    ]
)
```

```python
rbi_top10 = summary.sort_values("打点合計", ascending=False).head(10)

display(
    rbi_top10[
        ["球団", "打者名", "打席数", "RE24合計", "WPA合計", "打点合計", "本塁打合計", "1打席あたりRE24"]
    ]
)
```

```python
# ⑥ 打点とRE24の相関を確認
corr_rbi_re24 = summary["打点合計"].corr(summary["RE24合計"])

print("打点合計とRE24合計の相関係数:", corr_rbi_re24)
```

```python
corr_re24_wpa = summary["RE24合計"].corr(summary["WPA合計"])

print("RE24合計とWPA合計の相関係数:", corr_re24_wpa)
```
